# Lecture 7 - Artificial Intelligence    

The use of AI, especially generative AI has change the way that we work, including the world of programming.

## What is Copilot? 
Copilot is a gen AI model from Microsoft, whom can act as your pair programmer. It can suggest code, comments and tests your code, it even has some integrations in IDE's such as Visual Studio Code to edit or review your code as you write it. This should always be used as an assistant, the outputs should always be reviewed and tested before accepting it into your code. 

### How AI fits into a workflow
AI can be good for completing lines or functions as you are writing them, sometimes with a suggestion showing in the background as you type. AI can also be good for testing your program in many different ways quickly, where as a human may be thinking on the happy path and not think of all the possible errors could happen. 

### Prompting the AI
Writing clearly and leaving comments of what your intention/happy path is can help the AI understand what your intention is and can improve the suggestions they produce. You can also prompt the the AI with testing and potential issues and it will suggest test cases and edge cases.
When tests fail, you can show the code which fails and then ask the AI for a fix, when it has reviewed the code. It is important to review the code suggested and test it.

### Security and Secerts
It is vital that you NEVER ask the AI for handling API keys, passwords or any private data or use it's suggestions if mentions it as part of another prompt. Never give the AI any secret files, or have that information inside your IDE when the AI chat integration is enabled.

### Licensing
Code which has been generated by AI may resemble public sources. Check for potential licensing rules and polices before using it.

### Academic Honesty
When using AI for university work, make sure that the module policy allows the use of AI or you use it within the means it is permitted in. and if allowed, document the use of the AI in a readme or other notes and make sure you fully understand the code. 

### Productivity and Tools
Use small specific prompts for a targeted suggestion, use the AI response as a suggestion/starting point and use that to build from or review your current code.
