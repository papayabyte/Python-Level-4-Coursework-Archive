# Lecture 1 – Concerning Programming
### What is Programming and How Do We Learn to Program?

A program is a set of instructions that describe how to solve problems or perform a task. These instructions are written in a specialised language, which can be executed by a computer and understood by other programmers.

In simpler terms, programming is like writing a recipe or a list of directions but for a computer to follow.

## The Three Components of Programming
All programs share three fundamental components:

1. **Sequence** – Steps are carried out in order.
2. **Selection** – Some steps are included or removed, depending on certain conditions.
3. **Iteration** – Some steps are repeated until a condition is met, such as in loops.

### 1. Sequence Example
In programming, **sequence** simply means that instructions are executed one after another in order.

```python
print('Step 1: Gather ingredients')
print('Step 2: Mix ingredients')
print('Step 3: Bake the cake')
```

### 2. Selection Example
Selection allows the program to choose different paths based on a condition. This is usually done using **if**, **elif**, and **else** statements.

```python
age = 20
if age >= 18:
    print('You are an adult.')
else:
    print('You are a minor.')
```

### 3. Iteration Example
Iteration (or looping) means repeating a set of instructions until a certain condition is met.

```python
for i in range(1, 6):
    print(f'Step {i}: Repeat this action')
```

## Reading vs. Writing Code
Programming code is **read much more often than it is written** — the ratio is approximately **10:1**. This means clarity and readability are essential when writing programs.

> "Code is read more often than it is written." — a key principle of good programming practice.
