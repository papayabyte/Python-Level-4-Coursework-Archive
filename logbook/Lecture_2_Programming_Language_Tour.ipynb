# Lecture 2 – A Programming Language Tour.
## The History of Programming

> “You have to know the past to understand the present.” – Carl Sagan

To understand programming concepts today, we must first understand the history of programming and how it has evolved over time.

## Why Do We Have Computers?

Computers are excellent at carrying out repetitive tasks They are accurate, fast, and can work for long periods of time without becoming tired or bored. Humans, by contrast, are prone to fatigue, boredom, and mistakes when performing repetitive tasks.

A computer that carries out specific tasks accurately and quickly is very practical. When we can program computers to perform multiple different purposes, they become even more powerful and useful.

## First Generation Programming

The **Colossus computer** was programmed using plugs and switches, representing the very origins of computer programming.

- Programming at this stage was low-level and specific to the hardware.
- Each computer was programmed in a unique way, often built for a specific purpose.
- Programmers needed detailed knowledge of the hardware architecture.

## Second Generation Programming

Programming advanced with the introduction of **assembly languages**.

- These were short symbolic codes that corresponded to the basic operations of the computer.
- Still hardware-specific, often limited to particular CPUs.
- Programs were not portable between devices, rewriting was required for different architectures.

## Third Generation Programming

The next step was the rise of high-level programming languages which are closer to human language and easier to understand.

- Programs became portable across computers (e.g., Intel, AMD, Apple Silicon).
- Many languages emerged, serving different purposes:
  - **FORTRAN** – scientific and mathematical computing
  - **COBOL** – business and administrative computing
  - **C** and **Pascal** – procedural languages
  - **Java** and **C#** – object-oriented languages
  - **Haskell** and **Scala** – functional languages
- Some languages, like **Python**, are **multi-paradigm**, supporting different styles of programming.

## Compilers and Interpreters

There are two main ways to convert high-level languages into low-level machine instructions:

### Compilers
- Translate the entire program before it runs.
- Output is an executable file (e.g., .exe, .apk).
- Execution is fast, but compilation can take time.
- Examples: C, C++, Rust.

### Interpreters
- Execute the program line-by-line without prior compilation.
- Easier to test and modify quickly.
- Run slower than compiled languages.
- Examples: Python, PHP, JavaScript

## About Python

Python is a high-level, interpreted, general-purpose programming language, consistently ranking among the top three most popular languages worldwide.

### Why Python remains a leading language:
- Easy to learn and read.
- Used for web development, data analysis, AI, automation, and more.
- Supported by a massive global community.
- Still relevant decades after its creation.

Even older languages like FORTRAN and COBOL from the 1960s remain in use today — and Python will likely persist for many years to come.

Emerging languages such as Rust and Go (Golang) are gaining popularity, but Python remains one of the most versatile and accessible choices for programmers.
