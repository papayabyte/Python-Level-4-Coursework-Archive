# Be DRY not WET

## Good/Working Code

It is possible that when you start coding, your total focus is to get the thing to work, but the working aspect of the code is only a part of what we are trying to create.

### Indentation 

For Python, indentation is very important. It is written in python that the number of spaces (or a tab) in a indentation is four.
incorrect indentation make the code harder to read and understand.

### Reading and Writing 

As we have covered over and over, we know that the code is read way more than it is written. So, it make sense that you make your or someone elses job of reading this code is as easy as it can be, this is also when the use of DRY (Don't Repeat Yourself) comes into play.

It is important to keep it simple when writing the code, We will still need to be able to understand the code a long while after it has been written, so an approach into making this as easy as possible is a vital step in your programming planning. DRY (Don't Repeat Yourself) and the opposite which could be used to describe for not using it is WET (Write Everything Twice.)

### Why Don't Repeat Yourself 

In programming, it is the business of capturing the rules of real world in computer code, it makes a huge amount of sense to do this. 
Such as in databases. it is the business of capturing the structure of the real world in storing that efficiently. 
And finally in the use of UI design, it makes sense that the interactions of the real world and are intuitive and accessible.

A good piece of code (a function) should be doing exactly one thing, this function can then be recalled again for future use without having to rewrite the code.
Following this when testing the function it can either work or not and this can identify a single point of error and when the fix is made, it is fixed everywhere.

### Methodology  

- Understand the issues.
- Take a very small part of the issue and then solve it.
- Repeat, until all parts have been solved. 
- Combine these solutions of the smaller issues and then form them together as a complete solution.
