### Understanding a computer 

A very basic understanding of a computer is that a computer stores and processes values in binary, theses are stored in the computer's memory. Arithmetic operations are carried out inside the CPU. You can also find a small amount of memory in the CPU as well, but the majority used is the masses of the memory from the Random Access Memory sticks.

## Data Restrictions

All this explains some limitations in the values that can be stored. 



- The value of an integer is limited by the number of bits used to store one.
- The size of floating-point numbers is likewise limited, as is the precision of the number.
- The length of a string is limited by the amount of memory available.

There can be some performance issues with programming, the length of a string is limited to the amount of memory available, but with the amount of GBs in a modern machine, this will be very unlikely to be a practical issue.

## Computers are Dumb 

Computers will follow the instructions of a program, whether or not they actually make sense, so if you tell the computer to do the wrong thing it will still do it even if it is not the intended result.

Such as if you write a script to work out the area of a circle, but use the diameter instead of the radius, the computer will follow your instructions and claim the answer it reached it correct, but it will not be with the error in your program.

### Errors in programming 

 Which means that there are two types of errors in programming:

- Syntax errors are when the program does not follow the rules of the language, and fails to compile or run. 

- Logic (semantic) errors are when the program runs, but does not do what you intended.
