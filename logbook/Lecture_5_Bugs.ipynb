## What is a bug?

A bug is a mistake in a program which causes it to behave unexpectedly or incorrectly.
Bugs can occur from a variety of sources, including misunderstandings of the problem, incorrect assumptions and human error. 
This can range anywhere from a minor software glitch to a major system failure.

## Errors in programming 

There are two major types of errors in programming:

- Syntax Errors, are when the program does not follow the rules of the language and fails to compile or run. 
- Logic (Semantic) errors, are when the program runs, but does not do what you intended. 

### Semantic Errors

Logic errors can be defined into these possibilities:

- The program does not do what we intend (In one or more situations)
    - This one is the most problematic 
- The program does do what we intend, but this is not what it should be doing.
    - This one is not really the programmer's fault. It should not happen if the task has been specified properly. 

## Bugs
There are really two ways in which bugs manifest themselves:
- The program appears to run correctly, but produces incorrect results in some situations.
- The program fails (Crashes) in some situations.

### The chances of an error
Some error conditions are more likely than others.

Obvious errors should clearly be handled gracefully, But there come a point where this is not worthwhile. Consider the effort in programming, and the additional size of the program. This is such as the infamous BSoD screen (Blue Screen of Death screen) and similar fallbacks. 

## Code Reviews 

No code should ever enter a final system until it has been pear reviewed by a fellow developer, this applies to all developers of all seniority.
It is managed in Git, using branches. Developers work in their own branches, and can submit code for review before it is merged into the main branch. 
This is known as a pull request. 

### Pair Programming 

Another approach is to program in pairs, this would be two developers which us one keyboard and mouse (and other tools) and write their programs together, this can normally be done by one of the pair programming out load.

This is sometimes part of "Extreme Programming" where effectively it is a continuous code review. This process is quite expensive as two programmers are doing the work of one, but this can be very productive, especially when tackling tricky code.

## Making a fix

Sometimes a fix for a bug is obvious. Often there is more than one way to achieve the fix. 

- A properly engineered, long term, effective fix. 
- A quick and dirty hack that will solve the problem for the short-term. This one can sometimes be the most common choice, but this leads to technical debt. 

However the program has been fixed, it should be properly reviewed. 

## Debugging

### Dry Running 

Dry Running is the process of executing the code by hand and examining the results. This assumes that we know in what situations the code fails. 
It can be done in the head, or with the aid of a print out. 

### Using a Debugger 

A debugger is a tool which you can normally find in your IDE that will allow us to stop a program running a custom points, and then run it line by line.
So, we can see how the values in the variables change. 
This can be a extra skill to lean, but it can be very effective to your programming. 

### Peppering it with print statements

Add print statement throughout the program to examine data at various points.
This is quick and easy, and can be very effective, but the downside to the is that we have to remember to remove the print statements. 

### Building in a Debugger

Add a constant value that cab ve set as a boolean, if this set to True, it would execute suitably positioned print statements. 
These are easy to code into the program, but this can make the program much bigger than it needs to be.

### Logging

Code the program so that it logs what it is doing to the same file. These are common with web servers such as Apache and Nginx. 
Examining these log files can often show the cause of problems. 

### Logging Levels 
 
Code the program so that it logs what it is doing to some file, with the level of detail configurable.
Perusing the log file can often show the cause of problems. Tweak the setting if more is needed. 

### Taking a break

Go for a walk, a drink or read some news or watch a video. 

Very often taking a break which will take your mind away from the problem, can help you see the code from fresh eyes or an idea to fix the problem may come to mind when you're taking a break away from it.

### Asking a Friend 

Show your code to someone else, this can help because either the other person will see the problem or as you are explaining the code and the issue you are having that could make you aware of the issue at hand.
