# Lecture 4 - A Programmers Work
## Understanding a programmers' work

We will look at the tools of the trade of a programmer, what skills do they need 

### What does it take to be a programmer? 

#### A programmer will need a set of skills, such as having efficient use of their skill set.

This will include problem-solving skills, teamwork, communication and a familiarity with a "standard toolset". Sometimes the most favorable tools of a development aren't always possible.  

#### A Toolset of a developer

The tools of choice for a developer is hugely personal and it is important to use the best tools for the persons workflow, but with this being personal means that the best tool for Sam, may not be the same tool for Henry, so this is really dependant to the person's skills. 

People will like to use Microsoft Windows or use the other operating systems such as Apple's Mac OS/IOS or the Unix based operating systems of Linux, where these are very different to the way a developer works, some changes will be less different in the way they operate, such as VS Code or PyCharm which do very similar tasks but is manly a change of User Interface or how it communicates with different tools such as Git.

When you look at the details of the toolkit of a developer, you will see that all the little details are highly customised to be person's personal taste.

### What is inside a developer's toolkit

In a role as a developer, some aspects of the role you will be doing will be prescribed, such as; using git, github remotes/repos, CSCW. But it is most likely that some of the lower-level stuff would not mandated. 

It would be expected that you would be able to describe and justify your toolset of choice and be able to fully install it again in another workspace/laptop, or else you have customised the toolkit in a way which you don't even understand.

### Making the choice of a toolkit

The tools which you use will depend on:

- Availability / Cost
- Experience 
- Bigger Picture & Ecosystem 
- Suitability 
- Personal Preferences 
- Beliefs 

#### Customising a toolkit

Any tool (Operating System or IDE will be highly customisable)

- Personal Preference 
- Individuality 
- Eyesight Requirements 
- Special Needs
- Sense of Ownership

#### Standards of a toolkit

Since there are many ways of personalising you toolkit there are some standards that everyone should be able to use;

- Basic Windows File Management 
- Basic Linux or Mac File Management 
- Command-Line Git
- Basic Text Editors

### Programming in unison 

It is not very often where programming is done in isolation, it can become a very social activity. Any non-basic program or system will require several developers. 

#### Agile

Most modern day development follows a Agile approach, this is when the focus is directly on delivering value to the customers; solving problems and making money. 
This contrasts with the previous approaches of being design-planning heavy.

#### An Agile Day 

Most of these start with a standup meetings, where devs will report on their current work, such things might be mentioned:
- What I did yesterday.
- What I plan to do today. 
- What issues may be holding me back. 

Work to do will be managed in a issue ticketing software, and usually tasks will be arranged according to the value which they deliver. 
They will also have been scored according to the estimated effort required.
