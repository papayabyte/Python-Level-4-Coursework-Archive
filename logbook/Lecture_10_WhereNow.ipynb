## How do we get good at programming?

- Practice, Practice, Practice. You learn by doing 
- Read other people's work. Learn from other code, review it, read how it works, build on it.
- Learn from your previous mistakes.
- Keep doing it, and retrain yourself.

## The Developer

A developer's role is the produce software applications by writing, testing and maintaining code. This involves programming, but also many other skills such as; Collaboration, problem solve, communication are all essential skills for a successful developer. 

## The Tester

A testers role is the ensure the quality of the code written and the functionality of the software. The is creating test plans, identifying bugs and working with the devs to resolve these issues.

## Analyst and Designer 

Understanding the user's needs, and design software solutions that meets the needs of the user. They need to be able to understand the code, even though they don't write it, because they need to design something that is realistic and practical and communicating these goals with the devs and meeting the goals together. 

## Architect

Design the overall structure and framework of the software systems.
This is making high-level decisions about the technology stacks systems components and integration methods.

## Admin 

The systems admin's role is to manage and maintain a computer's systems and networks.
This is such as, installing software, configuring systems, monitoring performance and troubleshoot issues.

## The Manager

A manager's role is to oversee the development and delivery of software projects. This involves coordinating teams, time, resources and budget.

## Security Specialist 

A cyber security specialist's role is to protect a businesses computer systems and network from cyber threats. 
Keeping up to date on petitional issues, monitoring suspicious activity and responding to incidents.
