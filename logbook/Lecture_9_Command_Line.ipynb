# Lecture 9 - The Joy of Command Line

A lot of time, people spend countless time trying to do smaller tasks within the UI in a IDE, when the tasks could be done within a CMD/Terminal in one or very few steps.

## What is Command-Line?

Command-Line is the text based interface to your computer, a while ago it was the only interface a computer had, such as MSDOS.
Within the command line there are terminal commands which are typed into the terminal to carry out certain tasks, this can be tasks such as open a program, move or delete files.
Commands are interpreted by a shell, the commands which were used in the 70s and 80s are still used today on modern systems, they have stood the test of time, they will certainly still be there in many years to come.

## IDE and GUI

Graphical Interfaces always come and go over various versions of software and operating systems.
Going back to 1985, where the first GUI based windows appeared, it looks very different today compared to Windows 10 or 11, but the same commands ran on there, will work today.

There are some standards within these interfaces; 

- The left mouse button usually selects something.
- The right mouse button usually produces a menu.
- There is a way to move windows around, normally the top of the window.
- There is a way to close windows, normally a X somewhere at the top right.

Using these, is heavy on the metaphors, such as "desktop" is an example, Files are stored in "folders" which are opened in "Windows" on a "Desktop" these things aren't real in the sense we thing they do, they are good for a casual user for simple tasks and easy to learn, but this hides things that are very important to a programmer. A interface hides complexity, but not completely. And it hides complexity that you really need to understand.

Such as in mobile devices where a UI is fundamental to operating it, These try to hide everything, but in doing so they hide important details that you need to understand in order to use them effectively.

Within Windows it closely links a file with the application that opens it, This is convenient for casual users, and fine for simple tasks.
But there are dozens, if not more, apps we might want to use to open a Python program file. 

Windows also hides precisely where files are stored. Such as "My Documents" which you can click on from "Desktop" is not quite where you think it is. While point-and-click interfaces are convenient, the command-line offers greater flexibility and control.

### The difference between GUI and CMD

#### Difference between files

If you have two versions of the same program, which contains 1000 lines, how can you tell the differences between them? Within a GUI it could be hard to find a fast way to do this, but as for in the terminal. "Diff main_v1.py main_v2.py" and now it is done.

#### Deleting files

Lets say you have a folder with masses of log files, deleting just the .txt files in a GUI is time consuming, but in a terminal it is one command and done, such as 
"rm ~/focp-projects/project/src/log*.txt" and it is done.

### Basic File Operations 

    ls - list files in a directory
    cd - change directory
    cp - copy files
    mv - move (or rename) files
    rm - remove (delete) files
    mkdir - make directories
    rmdir - remove empty directories

### Finding Things 

    find - find files and directories
    grep - search file contents
    locate - quickly find files by name
    which - find the location of executables

### Working with Files 

    cat - display file contents
    less - view file contents one page at a time
    head - display the first few lines of a file
    tail - display the last few lines of a file
    nano / vim / emacs - text editors

### System Information 

    pwd - print working directory
    whoami - display current user
    df - display disk space usage
    top / htop - display running processes
    uname - display system information
