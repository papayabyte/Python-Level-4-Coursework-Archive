# Modern Error Handling 

## The Happy Path
A program specification usually describes what should happen if everything goes to plan. 
if all inputs are expected, all output devices are present, and so on. Making this the Happy Path.

### Coding the Happy Path
It is tempting to code the Happy Path first, and with good reason:

- It has good documentation.
- It is the most important part.
- it may be the most interesting part. (No one likes dealing with bugs!)
- It delivers the most value.

### Off the Path
We need to stray from the path at times, and we need to code what should happen for this.
A program should not just crash when something unexpected happens, either. At the same time your code should also not confidently produce incorrect results, which is the worst case situation.

When we are going to expend effort coding what we might think of as the less than happy path. We should never, ever underestimate the ability of the end user to abuse or misuse the best of all programs. 

###  The Happy Path (A Key Idea)

Most of the time, for most of the users, a program will follow the Happy Path, or at least, one of a few Happy Paths. We can then say after that everything else is an exception. 

## The Impact of Errors.
There are two possible outcomes in these error cases:
- The program will crash
- The program will display unreliable results. 

## Handling Errors: LBYL
One approach is to Look Before You Leap. 

Here, we check that everything is okay before we do anything, which may cause an error.
This means we need to validate all inputs and preconditions before proceeding.

An example would be checking that a string entered could plausibly represent an integer, this would be removing a int() converter from the input, because this code may cause an error, if the expected input isn't submitted. And checking that a string could plausibly represent a floating point number is significantly harder.

### The Happy Path
Most of the time, the program will follow the Happy Path. so, errors are intended to be an exception to the rule. We do need to handle them, but the code should not be dominated by these exceptions.

### Historical Note 

The Look Before You Leap approach is a common in older programming languages, which often had (what we think of as now) poor error handling mechanisms. Programmers had to be careful to avoid errors, rather than just handling them when the arose. In languages like C, this often led to code littered with check for errors, making it hard to read and maintain.

The LBYL approach is a bit like checking the weather forecast before going for a walk.
If the forecast predicts rain, you might choose to stay indoors. Or, you will only go out if you know you will stay dry.  
Alternatively, you might just go for the walk regardless of the forecast.

You might take a waterproof coat, or umbrella, just in case it rains.
If it does rain, you can put on your coat or open your umbrella. You deal with the situation as it arises. 

## Handling Errors: EAFP

An alternative approach is to easier to ask forgiveness than permission (EAFP).

Here, we just do what we want to do (the Happy Path), and handle any errors that arise.
This means we do not need to validate all inputs and preconditions before proceeding. 
This seems a strange approach at first, but it works well in practice, and produces much neater code. 

### EAFP or LYBL

In Python or any modern language, either are available. We could just use one, but in practice, we use both.

```python
class_size = int(input('How many students? '))
```

Most programmers would use both approaches here:

Use LBYL to check if the input is valid before processing it. <br>
Use EAFP to handle any unexpected errors that occur during processing.

### An Example of LBYL

```python
class_size = int(input('How many students? '))
if class_size <= 0:
    print('Invalid class size.')
else:
    # TODO: Proceed with processing
```

### An Example of EAFP

```python
try:
	class_size = int(input('How many students? '))
except ValueError:
	print('Invalid input. Please enter a number.')
else:
	if class_size > 0:
	    # TODO: Proceed with processing
	else:
	    print('Invalid class size.')   
```
